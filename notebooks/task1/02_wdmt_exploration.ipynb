```python
from pathlib import Path
import sys
import pandas as pd

start = Path.cwd().resolve()
root = next((path for path in (start, *start.parents) if (path / 'pyproject.toml').is_file()), None)
if root is None:
    raise FileNotFoundError('Cannot locate project root containing pyproject.toml')
for import_path in (root, root / 'src'):
    if str(import_path) not in sys.path:
        sys.path.insert(0, str(import_path))
from submission.parser import parse_all

records = parse_all([], str(root / 'data' / 'bronze' / 'wdmt.json'), '')
wdmt = pd.DataFrame(records).sort_values(['vehicle_id', 'recorded_at'])
wdmt.head()
```

```python
wdmt.assign(interval_s=wdmt.groupby('vehicle_id').recorded_at.diff().dt.total_seconds()).groupby('vehicle_id').interval_s.describe()
```
